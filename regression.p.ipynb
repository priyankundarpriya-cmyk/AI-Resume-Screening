```python
def predict_resume(resume_id=1, skills='Python, Pytorch, TensorFlow', experience=2, education=1, certification='AWS Certified', salary=50000, projects=3, ai_score=80, job_role='AI Researcher'):

    lst=[] #numpy list for starting the values

    lst=lst+[resume_id]

    skills=list(lb_skills.transform([skills]))
    lst=lst+skills

    lst=lst+[experience]

    lst=lst+[education]

    certification=list(lb_certifications.transform([certification]))
    lst=lst+certification

    lst=lst+[salary,projects,ai_score]

    if job_role=='AI Researcher':
        lst=lst+[1,0,0,0]

    elif job_role=='Cybersecurity Analyst':
        lst=lst+[0,1,0,0]

    elif job_role=='Data Scientist':
        lst=lst+[0,0,1,0]

    elif job_role=='Software Engineer':
        lst=lst+[0,0,0,1]

    result=model.predict([lst])

    if result==[1]:
        return "Candidate is selected for hiring"

    elif result==[0]:
        return "Candidate is not selected for hiring"
```

```python
predict_resume()
```

```python
resume_id=int(input("Enter Resume ID: "))

skills=input("Enter skills: ")

experience=float(input("Enter your experience in years: "))

education=int(input("Enter education level(0/1/2/3): "))

certification=input("Enter certification: ")

salary=float(input("Enter salary expectation: "))

projects=int(input("Enter number of projects: "))

ai_score=float(input("Enter AI score(0-100): "))

job_role=input("Enter job role(AI Researcher/Cybersecurity Analyst/Data Scientist/Software Engineer): ")

result=predict_resume(resume_id,skills,experience,education,certification,salary,projects,ai_score,job_role)

print(result)
```
